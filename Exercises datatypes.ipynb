````python
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}

# 1. Join A and B
print("A union B:", A | B)

# 2. Find A intersection B
print("A intersection B:", A & B)

# 3. Is A subset of B?
print("Is A subset of B?", A.issubset(B))

# 4. Are A and B disjoint?
print("Are A and B disjoint?", A.isdisjoint(B))

# 5. Join A with B and B with A
print("A | B:", A | B)
print("B | A:", B | A)

# 6. Symmetric difference
print("Symmetric difference:", A ^ B)

# 7. Delete sets completely
del A, B
````

### Exercises ➞ <span class='label label-default'>Level 2</span>

1. The following is a list of 10 students ages:
```py
ages = [19, 23, 19, 25, 21, 20, 25, 26, 25, 24]
```
    
- Sort the list and find the min and max age
- Add the min age and the max age again to the list
- Find the median age (one middle item or two middle items divided by two)
- Find the average age (sum of all items divided by their number )
- Find the range of the ages (max - min)
- Compare the value of (min - average) and (max - average), use **[abs()]** method

2. Find the middle country(ies) in the **[countries list]()**
3. Divide the countries list into two equal lists if it is even if not one more country for the first half.
4. **`['India', 'Russia', 'China', 'Finland', 'Sweden', 'Norway', 'Denmark']`**. Unpack the first three countries and the rest as scandic countries.



---

````python
ages = [19, 23, 19, 25, 21, 20, 25, 26, 25, 24]

sorted_ages = sorted(ages)
min_age = min(sorted_ages)
max_age = max(sorted_ages)

# Add min and max again
ages_plus = sorted_ages + [min_age, max_age]
print("Sorted ages:", sorted_ages)
print("Min age:", min_age)
print("Max age:", max_age)
print("Ages with min and max added:", ages_plus)
````

````python
# Median
if len(sorted_ages) % 2 == 1:
    median = sorted_ages[len(sorted_ages) // 2]
else:
    mid1 = sorted_ages[len(sorted_ages) // 2 - 1]
    mid2 = sorted_ages[len(sorted_ages) // 2]
    median = (mid1 + mid2) / 2
print("Median:", median)
````

````python
# Average
average = sum(sorted_ages) / len(sorted_ages)
print("Average:", average)
````

````python
# Range
age_range = max_age - min_age
print("Range:", age_range)
````

````python
# Compare absolute differences
print("abs(min - average):", abs(min_age - average))
print("abs(max - average):", abs(max_age - average))
````

````python
# Middle country
countries = ["India", "Russia", "China", "Finland", "Sweden", "Norway", "Denmark"]
mid_country = countries[len(countries) // 2]
print("Middle country:", mid_country)

# Divide list into two parts
if len(countries) % 2 == 0:
    first_half = countries[:len(countries) // 2]
    second_half = countries[len(countries) // 2:]
else:
    first_half = countries[:len(countries) // 2 + 1]
    second_half = countries[len(countries) // 2 + 1:]
print("First half:", first_half)
print("Second half:", second_half)
````

````python
# Unpack first three countries and the rest
country_list = ["India", "Russia", "China", "Finland", "Sweden", "Norway", "Denmark"]
first, second, third, *scandic_countries = country_list
print("First three:", first, second, third)
print("Scandinavian countries:", scandic_countries)
````

````python
# 1. Empty dictionary
bird = {}

# 2. Add keys
bird["name"] = "Parrot"
bird["color"] = "Green"
bird["breed"] = "African Grey"
bird["legs"] = 2
bird["age"] = 2
print("bird:", bird)
````

````python
# 3. Create student dictionary
student = {
    "first_name": "John",
    "last_name": "Doe",
    "gender": "Male",
    "age": 21,
    "marital_status": "single",
    "skills": ["Python", "JavaScript"],
    "country": "Egypt",
    "city": "Cairo",
    "address": "Nasr City"
}
````

````python
# 4. Length
print("Length of student:", len(student))

# 5. Get value of skills and check data type
skills_value = student["skills"]
print("skills:", skills_value)
print("Type of skills:", type(skills_value))
````

````python
# 6. Modify skills
student["skills"].extend(["SQL", "Django"])
print("Updated skills:", student["skills"])
````

````python
# 7. Get keys as list
keys_list = list(student.keys())
print("Keys:", keys_list)

# 8. Get values as list
values_list = list(student.values())
print("Values:", values_list)
````

````python
# 9. Change dictionary to list of tuples
items_list = list(student.items())
print("Items as tuples:", items_list)

# 10. Delete one item
del student["address"]
print("After deleting address:", student)

# 11. Delete the dictionary
del student
# print(student)  # would error because it is deleted
````
